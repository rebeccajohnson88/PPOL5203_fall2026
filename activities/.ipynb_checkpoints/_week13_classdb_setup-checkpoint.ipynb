Re-running is safe: the tables are replaced each time.

````python
# Install the Postgres driver into this notebook's Python if it's missing
import importlib.util, subprocess, sys
for pkg, pip_name in [("sqlalchemy", "sqlalchemy"), ("psycopg2", "psycopg2-binary")]:
    if importlib.util.find_spec(pkg) is None:
        subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", pip_name])

from pathlib import Path
import os

import pandas as pd
from sqlalchemy import create_engine, text
````

## 1. Read the connection strings from `.env.local`

````python
def load_env_local(filename=".env.local"):
    """Find .env.local in this folder or a parent folder and load it into os.environ."""
    for folder in [Path.cwd(), *Path.cwd().parents]:
        path = folder / filename
        if path.exists():
            for line in path.read_text().splitlines():
                line = line.strip()
                if line and not line.startswith("#") and "=" in line:
                    key, value = line.split("=", 1)
                    os.environ.setdefault(key.strip(), value.strip().strip('"').strip("'"))
            return folder
    raise FileNotFoundError(f"Could not find {filename} in {Path.cwd()} or any parent folder")

COURSE_ROOT = load_env_local()
DATASETS = COURSE_ROOT / "datasets"

# Print only whether each variable is set, never its value
for var in ["DATABASE_URL", "STUDENT_DATABASE_URL"]:
    print(var, "found" if os.environ.get(var) else "MISSING")
````

## 2. Load the CSVs

````python
TABLES = {
    "caseinit": "init_fordb.csv",
    "divert": "divert.csv",
}
DATE_COLS = ["received_date", "referral_date", "diversion_closed_date"]

def load_csv(filename):
    df = pd.read_csv(DATASETS / filename)
    # Lowercase column names so students never need "double quotes" in SQL
    df.columns = [c.lower() for c in df.columns]
    for col in DATE_COLS:
        if col in df.columns:
            df[col] = pd.to_datetime(df[col], format="%m/%d/%Y %I:%M:%S %p", errors="coerce")
    return df

frames = {table: load_csv(f) for table, f in TABLES.items()}
for table, df in frames.items():
    print(f"{table}: {df.shape[0]:,} rows x {df.shape[1]} columns")
````

## 3. Write the tables to Neon (as the owner)

````python
owner_engine = create_engine(os.environ["DATABASE_URL"], pool_pre_ping=True)

for table, df in frames.items():
    df.to_sql(table, owner_engine, if_exists="replace", index=False,
              method="multi", chunksize=2000)
    print(f"wrote {table}")

# Belt and braces: make sure the student login can read the (re)created tables
with owner_engine.begin() as conn:
    conn.execute(text("GRANT USAGE ON SCHEMA public TO student"))
    conn.execute(text("GRANT SELECT ON ALL TABLES IN SCHEMA public TO student"))
````

## 4. Check everything as a student would see it

````python
student_engine = create_engine(os.environ["STUDENT_DATABASE_URL"], pool_pre_ping=True)

for table in TABLES:
    n = pd.read_sql(f"SELECT COUNT(*) AS n FROM {table}", student_engine)["n"][0]
    print(f"{table}: {n:,} rows visible to the student login")

pd.read_sql("SELECT * FROM caseinit LIMIT 5", student_engine)
````

````python
# The student login should NOT be able to change anything
try:
    with student_engine.begin() as conn:
        conn.execute(text("DELETE FROM divert"))
    print("PROBLEM: the student login was able to delete rows")
except Exception as e:
    print("Good: write blocked for student login ->", type(e).__name__)
````
