## Activity 3: strings

- The previous activities contained strings as elements of the lists/dictionaries
- Let's now practice some string methods
- See the string below from recent McCourt news 
- See [here](https://www.w3schools.com/python/python_ref_string.asp) for string methods

```python
mcc = """
mcc = "There’s been a flurry of activity around the building,
and I’m here for it, as my kids would say.  We are exceptionally 
proud of, and grateful for, our beautiful physical space here at
125 E Street and last week, we became even prouder."
"""
```

A. Determine if the word "Washington" is present in the string. Determine if the word "week" is in the string.

```python
"Washington" in mcc
```

```python
"week" in mcc
```

B. Count how many times the word "grateful" appears

```python
# count
mcc.count("grateful")
```

```python
# where
mcc.index("grateful")
```

```python
grateful_index = mcc.index("grateful")
```

```python
# print the word 
# we subset the string to the index (start)
# and then the next position + the len of the string 
mcc[grateful_index:grateful_index+len("grateful")]
```

C. Convert the entire string to upper case

```python
mcc.upper()
```

D. Replace the phrase "We" with "I" throughout the string

```python
mcc.replace("We", "I")
```

E. Run the below code to strip any punctuation (like commas, exclamation marks, etc.) from the string - next week, we'll be covering `for

```python
# Define a string containing some punctuation marks.
punctuation = ",'.!"

# Iterate over: we will see this later. 
for char in punctuation:
    mcc = mcc.replace(char, "")
    
print(mcc)    
```
