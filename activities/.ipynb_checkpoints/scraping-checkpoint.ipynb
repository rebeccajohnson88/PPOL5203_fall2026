3. Use Inspect or SelectorGadget to find CSS selectors for, and then extract:

**Hints:**

```python
## get access to the website
url = "https://www.georgetown.edu/news/chinese-language-joy-of-learning/"
page = requests.get(url)

## check if you got a connection
page.status_code
```

```python
## create a bs object
soup = BeautifulSoup(page.content, 'html.parser')
```

```python
## first attempt at the text: all paragraphs
all_p = soup.select("p")
len(all_p)

## the last one is the page footer, not the story
all_p[-1].get_text()
```

```python
## more specific css selector: paragraphs inside the article body
body_p = soup.select(".c--article-body-text p")
len(body_p)

## clean and join together with string methods
story_text = "\n".join([p.get_text().strip() for p in body_p])
print(story_text[0:500])
```

```python
## title: the h1 inside the page title field
title_tag = soup.select(".f--page-title h1")[0]
title_tag
```

Notice the `<span class="visually-hidden">Title: </span>` inside the tag: it's text for screen readers that isn't displayed on the page, but `.get_text()` still picks it up. Two ways to remove it:

1. Use string methods to remove the label after extracting the text
2. Remove the hidden span from the html before extracting the text (`.decompose()`)

```python
## option 1: string methods
title_tag.get_text().replace("Title:", "").strip()

## option 2: remove all visually-hidden spans from the soup
## (this modifies the soup, so it also cleans up the date and author)
for hidden in soup.select(".visually-hidden"):
    hidden.decompose()

story_title = soup.select(".f--page-title h1")[0].get_text().strip()
story_title
```

```python
## date published
story_date = soup.select(".date")[0].get_text().strip()
story_date

## author
story_author = soup.select(".author")[0].get_text().strip()
story_author
```

```python
## store in a one-row dataframe
one_article = pd.DataFrame({"url": url,
                            "title": story_title,
                            "date": story_date,
                            "author": story_author,
                            "text": story_text}, index = [0])
one_article
```

## B. Write a function to scrape one article

Write a function called `scrape_gu_article` that:

- Takes one input: the url of a Georgetown news article
- Returns a dictionary with the url, title, date, author, and text of the article

Then test your function on these two articles:

- https://www.georgetown.edu/news/chinese-language-joy-of-learning/
- https://www.georgetown.edu/news/board-directors-fall-2026/

**Hint:** not every article lists an author. What happens to your code from Part A when there's no `.author` element on the page? Write your function so it returns `None` for the author in that case rather than throwing an error.

```python
def scrape_gu_article(url):
    """
    Scrapes one article from the Georgetown news site

    Input:
        url (str): url of a Georgetown news article
    Returns:
        dict with the url, title, date, author, and text of the article
    """
    ## get access to the website
    page = requests.get(url)

    ## check if you got a connection
    if page.status_code != 200:
        print(f"Status {page.status_code} for {url}")
        return None

    ## create a bs object
    soup = BeautifulSoup(page.content, 'html.parser')

    ## remove text that's hidden on the page (e.g., "Title: ")
    for hidden in soup.select(".visually-hidden"):
        hidden.decompose()

    ## title
    story_title = soup.select(".f--page-title h1")[0].get_text().strip()

    ## date
    story_date = soup.select(".date")[0].get_text().strip()

    ## author: not every article has one, so check before indexing
    author_tags = soup.select(".author")
    if len(author_tags) > 0:
        story_author = author_tags[0].get_text().strip()
    else:
        story_author = None

    ## text
    body_p = soup.select(".c--article-body-text p")
    story_text = "\n".join([p.get_text().strip() for p in body_p])

    return {"url": url,
            "title": story_title,
            "date": story_date,
            "author": story_author,
            "text": story_text}
```

```python
## test on an article with an author
test_1 = scrape_gu_article("https://www.georgetown.edu/news/chinese-language-joy-of-learning/")
{key: value[0:100] if isinstance(value, str) else value for key, value in test_1.items()}

## test on an article without an author
test_2 = scrape_gu_article("https://www.georgetown.edu/news/board-directors-fall-2026/")
{key: value[0:100] if isinstance(value, str) else value for key, value in test_2.items()}
```

## C. Write a function to collect article urls

Go to https://www.georgetown.edu/news/all-news/ and write a function called `collect_gu_links` that:

- Takes one input: the url of the news listing page
- Returns a list of the urls for the articles listed on that page

Then run your function and look at the urls it returns.

**Hints:**

- Use Inspect or SelectorGadget on one of the article headlines to find a css selector for the headline links, then pull the `href` attribute from each link
- Look closely at the urls: do they all point to pages on `www.georgetown.edu/news/`? Why might links to other Georgetown sites (e.g., the medical center or business school) cause problems for the function you wrote in Part B? Modify your function to keep only the links that will work with `scrape_gu_article`

```python
## step 3 from our routine: collect one realization of the data
listing_url = "https://www.georgetown.edu/news/all-news/"
page = requests.get(listing_url)
soup = BeautifulSoup(page.content, 'html.parser')

## headline links are inside the cta title field
headline_links = soup.select(".f--cta-title a")
len(headline_links)
headline_links[0]

## pull the href attribute
all_links = [link["href"] for link in headline_links]
all_links
```

Some of the links go to other Georgetown sites (e.g., `lombardi.georgetown.edu`, `msb.georgetown.edu`). Those sites are built differently, so the css selectors in `scrape_gu_article` may not find anything there --- and we can't assume they will. We'll keep only the links on `www.georgetown.edu/news/`.

```python
## step 4: generalize into a function
def collect_gu_links(listing_url):
    """
    Collects article urls from the Georgetown news listing page

    Input:
        listing_url (str): url of the news listing page
    Returns:
        list of unique article urls on www.georgetown.edu/news/
    """
    ## get access to the website
    page = requests.get(listing_url)
    soup = BeautifulSoup(page.content, 'html.parser')

    ## pull the href attribute from each headline link
    all_links = [link["href"] for link in soup.select(".f--cta-title a")]

    ## keep only links on the main news site that our scraper works for
    gu_links = [link for link in all_links
                if re.match(r'^https://www\.georgetown\.edu/news/', link)]

    ## remove duplicates while keeping the order
    gu_links = list(dict.fromkeys(gu_links))

    return gu_links
```

```python
gu_links = collect_gu_links("https://www.georgetown.edu/news/all-news/")
len(gu_links)
gu_links
```

## D. Combine the two functions to scrape all articles on the listing page

1. Use `collect_gu_links` to get the article urls
2. Loop over the urls and use `scrape_gu_article` to scrape each one
    - Pause for a random amount of time between requests to be kind to the server
    - Use `try` and `except` so that one broken page doesn't stop the whole loop; store any urls that fail along with their error
3. Combine the results into a pandas DataFrame with one row per article

**Challenge:** run your loop on *all* the headline links from Part C (before filtering to `www.georgetown.edu/news/`). Which urls end up in your list of errors, and what error do they produce?

```python
## step 1: collect the links
gu_links = collect_gu_links("https://www.georgetown.edu/news/all-news/")

## step 2: loop through and scrape each article
scraped_data = []
list_of_errors = []

for url in gu_links:

    ## check what is going on
    print(url)

    ## scrape the content
    try:
        one_result = scrape_gu_article(url)
        if one_result is not None:
            scraped_data.append(one_result)
    except Exception as e:
        list_of_errors.append([url, e])

    ## put the system to sleep for a random draw of time (be kind)
    time.sleep(random.uniform(.5, 2))
```

```python
## step 3: organize as a pandas data frame
gu_news = pd.DataFrame(scraped_data)
gu_news.shape
gu_news[['title', 'date', 'author']]

## any errors?
list_of_errors
```

```python
## quick checks on the scraped data:
## how long is each article, and how many list an author?
gu_news['n_words'] = gu_news.text.str.split().str.len()
gu_news[['title', 'n_words']].sort_values(by = 'n_words', ascending = False)

gu_news.author.notnull().mean()
```

### Challenge solution

```python
## run the loop on all the headline links (including other Georgetown sites)
scraped_all = []
errors_all = []

for url in all_links:
    try:
        one_result = scrape_gu_article(url)
        if one_result is not None:
            scraped_all.append(one_result)
    except Exception as e:
        errors_all.append([url, e])

    time.sleep(random.uniform(.5, 2))

## if a page on another Georgetown site doesn't have a ".f--page-title h1" element,
## soup.select() returns an empty list and indexing it with [0] raises an IndexError;
## compare the urls here with the ones that were scraped successfully
errors_all
```
