## Activity 1: For loops 

```python
## start with these objects
list_exercise = ["Giants", "Commanders", "Eagles", "Jets"] 
sports_dict = {'Team': ["Giants", "Commanders", "Eagles", "Jets"],
              'City': ["Dallas", "DC", "Philadelphia", "NYC"],
              'Week1': [1, 0, 1, 1]} 
```

A. Using a `for` statement, iterate over each element in `list_exercise` and print a lowercase version of the element. Print the `list_exercise` list after the loop and check to make sure the elements have not been modified 

```python
for element in list_exercise:
    print(element.lower())
```

B. Using a `for` statement, iterate over each element in `list_exercise` and print the following statement: "The team is called the [insert name of team]"

C. Run the following code to iterate over the values in `sports_dict` to print the following statement: "The [Insert name of team] play in [Insert city of team]"

**Note**: here, we're iterating using the positions of the elements in each of the dictionary's lists. Later, we'll learn a more efficient way to do this where we don't need to separate out the lists

```python
teams = sports_dict['Team']
cities = sports_dict['City']
for i in range(len(teams)):
    print("The " + teams[i] + " play in " + cities[i])
    
```

D. Use an `if` `else` statement within a `for` loop to iterate over the list in the `Week1` key of `sports_dict`. If the week 1 results takes the value of 1, print "[Insert team name] won!" Otherwise, print "[Insert team name] lost :("

**Note**: because we only have two possible outcomes, we can skip the `elif` part from the slides

```python
results = sports_dict['Week1']
for i in range(len(teams)):
    if results[i] == 1:
        print(teams[i] + " won!")
    else:
        print(teams[i] + " lost :(")
```

## Activity 2: while loop

A. Initialize a value of `x = 3`. Use the `x-=1` syntax to remove 1 from x with each iteration of the while loop. Stop the loop before x falls below 0

```python
x = 3
while x >= 0:
    print(x)
    x -= 1
```

A. Define a function called `go_team` that takes in a single argument called `team_name`, a string for team, and returns the phrase ("Go [insert name of team]!")

Execute the function using string "Commanders"


```python
def go_team(team_name):
    '''
    Creates a string to cheer for a team
    team_name: str object
    output: returns the cheering phrase
    '''
    phrase = "Go " + team_name + "!"
    return phrase 

go_team(team_name = "Commanders")
```

B. Define a function called `go_team_city` that takes in two arguments:
- `team_name`: string for name of team
- `team_city`: string for city of team
    
And returns the phrase: "Go [insert name of team] from [insert name of city]!"
    
Execute the function using Commanders as the team name and DC as the city

```python
def go_team_city(team_name, team_city):
    '''
    Creates a string to cheer for a team from a specific city
    team_name: str object
    team_city: str object
    output: returns the cheering phrase
    '''
    phrase = "Go " + team_name + " from " + team_city + "!"
    return phrase 

go_team_city(team_name = "Commanders", team_city = "DC")
```

C. Execute that same function but instead of feeding team_name and team_city as named arguments, feed them as positional arguments

## Activity 5: Generators 

A. Use the `zip` function to zip together the lists of teams (`sports_dict['Team']`) and results (`sports_dict['Week1']`). Store this zip object as team_w_results

```python
team_w_results = zip(sports_dict['Team'], sports_dict['Week1'])
```

B. Using a for loop, iterate over the zipped object and print "[Insert team name]  won!" if the results is equal to 1; else print "[Insert team name] lost :("

**Note**: this is same task as Activity 1, Exercise D but you should be able to use simpler syntax that avoids the range generator

```python
for team, result in team_w_results:
    if result == 1:
        print(team + " won!")
    else:
        print(team + " lost :(")
```
