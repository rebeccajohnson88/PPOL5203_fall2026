# Activity 2: selecting and filtering 

A. Subset the data to the State == `California`. Store the resulting dataframe as `ca_df` and use the `shape` command to print the dimensions 

```python
ca_df = df[df.State == "California"].copy()
ca_df
```

B. Use list comprehension and the `columns` attribute to create a list of columns, called `cols_keep` that excludes the following two columns: 
- `State`
- `Census_Division`

Print the list. It should contain columns for Race, Year, and uninsured_rate

```python
cols_keep = [col for col in ca_df.columns if col not in ["State", "Census_Division"]]
cols_keep
```

C. Use the `cols_keep` list to create a new dataset, `ca_df_clean` that removes those two columns. Print the head

```python
ca_df_clean = ca_df[cols_keep].copy()
```

```python
ca_df_clean.head()
```

D. Write a user-defined function, `subset_onerace` that takes in three arguments:
- A string for a racial/ethnic group
- A string representing the name of the race/ethnicity column
- A dataframe

And subset the data to rows of data for that race/ethnicity

```python
def subset_onerace(one_race, race_colname, df):
    """
    one_race: string for race/ethnicity category
    race_colname: str with name of the race/ethnicity column
    df: dataset
    Returns: dataset subsetted to that race/ethnicity 
    """
    # one approach
    df_sub = df[df[race_colname] == one_race]
    # could also use query and the like approach if want to match more incomplete category names
    return df_sub
```

E. Use that user-defined function to return the rows for `American Indian or Alaska Native`, ordered by year from lowest to highest. Has the uninsurance rate among children gone up or down over time?

```python
ca_df_clean.Race.value_counts()
```

```python
subset_onerace(one_race = "American Indian or Alaska Native",
              race_colname = "Race",
              df = ca_df_clean).sort_values(by = 'Year')
```

F. Using list comprehension, iterate over the unique race/ethnicity groups in `ca_df_clean` and apply that function to create a subset of rows ordered by year for that group; the result will be a list of dataframes

Print the result; what patterns do you notice?

```python
separate_groups = [subset_onerace(one_race = x,
                                  race_colname = "Race",
                                  df = ca_df_clean).sort_values(by = 'Year')
                  for x in ca_df_clean.Race.unique()]
```

# Activity 3: creating and recoding variables 

A. Going back to the full dataset, `df`, create a new variable in the dataset, `race_simplified`, that:

- Converts the string to lowercase
- Replaces spaces with _ 

Use `pd.crosstab` with the two variables to make sure the recoding worked 

```python
df['race_simplified'] = df.Race.str.replace(" ", "_").str.lower()
pd.crosstab(df.Race, df.race_simplified)
```

B. In the full dataset, use `np.mean` to find the nationwide average uninsurance rate. Create a variable, `ur_demean` that subtracts that national average from each individual `uninsured_rate`

Use .describe() to confirm the de-meaning worked (the overall variable mean should be very close to zero and min and max should change)

```python
# display option rounding to 4 sig difits
pd.options.display.float_format = '{:.4f}'.format
ur_mean = np.mean(df.uninsured_rate)
df['ur_demean'] = df.uninsured_rate - ur_mean

df.describe()
```

C. In the full dataset, use `np.select` to create a new variable, `type_statename` that codes state names into two categories:
    
    - two_words: if the name has multiple words (eg North Carolina, New Jersey, etc)
    - one_word: if the name has a single word
        
    Hint: you can do this by checking whether the name string contains any spaces, comparing the length of the string to the number of alpha characters, or another approach

Spot check your work using the `sample` command

```python
conditions = [df.State.str.contains(" "),
             ~df.State.str.contains(" ")]
choices = ["two_words",
          "one_word"]

df['type_statename'] = np.select(conditions, choices, default = np.nan)
df.sample(10)
```

D. Use `groupby` and the built-in `idxmax` aggregation function to:
    
- Group by race/ethnic group (`Race`)
- Find the year and state with the highest uninsurance rate for that race/ethnic group

Hint: there are multiple approaches for how to pull the columns tied to the row with the maximum uninsurance rate, including .loc and sort_values()

```python
df.loc[df.groupby('Race')['uninsured_rate'].idxmax(),
       ['Race', 'State', 'Year', 'uninsured_rate']]
```

E. Use `groupby`, `agg`, and a lambda function to:

- Group by race/ethnic group (`Race`)
- Find the interquartile range of uninsurance rates for each race/ethnicity group across states and years. Which groups vary the most? 

```python
df.groupby('Race').agg(ur_iqr = 
                      ('uninsured_rate',
                      lambda x: x.quantile(0.75) - x.quantile(0.25)))
```
