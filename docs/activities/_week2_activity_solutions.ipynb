# Activity 1: cloning and editing a repo 

1. Create a new public repo using GitHub.com, calling it ppol5203_f26_myactivities 
2. Clone the repo locally
3. Create a code/ subdirectory
4. Move this python notebook to the code/subdirectory, naming it "week2_activity.ipynb"
5. Add, commit, and push the changes with the following message "Code changes"
6. Create an output/ subdirectory

```python
## I'm putting the steps I ran in command line here
## I added a cell by going to Insert -> insert cell below
## I then clicked on cell and selected "Code" under "Cell type"

## cloning step (your url will differ slightly)
git clone https://github.com/rebeccajohnson88/ppol5203_f26_myactivities.git

## navigating into my cloned folder and making a code subdirectory
## navigating into that directory
cd ppol5203_f26_myactivities
mkdir code/ 
cd code/

## moving python notebook into my code subdirectory (your mv)
## path will differ depending on where your notebook was downloaded
## i'm using the . as the destination folder for the move command
## since im in the code subdirectory already
mv ../../class_repo/week2_activity.ipynb .

## add, commit, and push that file 
## (can also push the entire code subdirectory)
git add week2_activity.ipynb
git commit -m "Code changes"
git push 

## navigating out of the code folder and making an output folder
cd ../
mkdir output 

## navigating into that output directory 
## and using touch to create a blank file
cd output
touch example.txt

## add committing and pushing
git add example.txt
git commit -m "Output updates"
git push

## final step was on the web platform 
```
