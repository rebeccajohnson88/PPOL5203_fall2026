# Activity 4: grouping and aggregation

A. Group the dataset by `Census_Division`. Use a for loop to print the group names  

```python
## your code here
cd_groups = df.groupby('Census_Division')
for group, data in cd_groups:
    print(group)
```

B. Use `get_group` to extract the Pacific region group and find the mean `uninsured_rate` for states/years/racial/ethnic groups in that group

```python
## your code here
pacific_group = cd_groups.get_group('Pacific')
pacific_group.uninsured_rate.mean()
pacific_group['uninsured_rate'].mean()
```

C. Going back to the full dataset, group by `Census_Division` and find the mean uninsurance rate within that group across all states/years. Was the Mountain region higher or lower than other regions in terms of its child uninsurance rate? Which region has the lowest uninsurance rate?

```python
## your code here
df.groupby('Census_Division')['uninsured_rate'].mean().reset_index().sort_values(by = 
                                                        'uninsured_rate')
```

```python
df.groupby('Census_Division').agg({'uninsured_rate':
                                   ['mean', 'median']}).reset_index()
```

D. Use `groupby` and the built-in `idxmax` aggregation function to:
    
- Group by race/ethnic group (`Race`)
- Find the year and state with the highest uninsurance rate for that race/ethnic group

Hint: there are multiple approaches for how to pull the columns tied to the row with the maximum uninsurance rate, including .loc and sort_values()

```python
## your code here
index_max = df.groupby('Race')['uninsured_rate'].idxmax()
index_max

df.loc[index_max, ['Race', 'State', 'Year', 'uninsured_rate']]
```

E. Use `groupby`, `agg`, and a lambda function to:

- Group by race/ethnic group (`Race`)
- Find the interquartile range of uninsurance rates for each race/ethnicity group across states and years. Which groups vary the most? 

```python
## your code here 
df.groupby('Race').agg(ur_iqr = ('uninsured_rate',
                                lambda x: x.quantile(0.75) - x.quantile(0.25)))
```
